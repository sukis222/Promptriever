Ноутбук загружает InfoSearch с Hugging Face и выводит его формат. Он нужен для проверки запросов, корпуса и qrels перед оценкой.


```python
!pip install -q datasets huggingface_hub

import os, json
from collections import Counter, defaultdict
from datasets import load_dataset
from huggingface_hub import snapshot_download

DATASET = "EIT-NLP/InfoSearch"
LOCAL_DIR = snapshot_download(repo_id=DATASET, repo_type="dataset")
TASKS = ["Audience-v1", "Format-v1", "Keyword-v1", "Language-v1", "Length-v1", "Source-v1"]

print("Dataset:", DATASET)
print("Local dir:", LOCAL_DIR)
print("Tasks:", TASKS)

for task in TASKS:
    base = os.path.join(LOCAL_DIR, task)
    files = {
        "queries": os.path.join(base, "queries.jsonl"),
        "corpus": os.path.join(base, "corpus.jsonl"),
        "qrels_og": os.path.join(base, "qrels_og", "test.jsonl"),
        "qrels_changed": os.path.join(base, "qrels_changed", "test.jsonl"),
        "qrels_reversed": os.path.join(base, "qrels_reversed", "test.jsonl"),
    }
    print("\n" + "=" * 100)
    print(task)
    for name, path in files.items():
        n = 0
        first = None
        with open(path, "r", encoding="utf-8") as f:
            for line in f:
                obj = json.loads(line)
                if first is None:
                    first = obj
                n += 1
        print(name, n)
        print(json.dumps(first, ensure_ascii=False, indent=2))
```
